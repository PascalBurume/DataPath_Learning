## L7.4 — 9-step EDA checklist `[Expert]`

```python
import pandas as pd, numpy as np, seaborn as sns, matplotlib.pyplot as plt
np.random.seed(42)
df = pd.read_csv('housing_prices.csv')
df.info()
```

## L7.2 — Pearson vs Spearman `[McKinney Ch5 p.175–180]`

```python
num = df.select_dtypes('number')
p = num.corr(method='pearson')
s = num.corr(method='spearman')
fig, ax = plt.subplots(1, 2, figsize=(12, 4))
sns.heatmap(p, ax=ax[0], annot=False, cmap='coolwarm', center=0); ax[0].set_title('Pearson')
sns.heatmap(s, ax=ax[1], annot=False, cmap='coolwarm', center=0); ax[1].set_title('Spearman')
plt.tight_layout()
```

## L7.7 — Bivariate visual EDA `[McKinney Ch9 p.310–335]`

```python
sns.pairplot(num.sample(min(500, len(num))).head(), diag_kind='kde')
```
