## L8.1 — Figure / Axes object model `[McKinney Ch9 p.270–290]`

```python
import pandas as pd, numpy as np, matplotlib.pyplot as plt
np.random.seed(42)
df = pd.read_csv('air_quality_daily.csv', parse_dates=['date'])

fig, ax = plt.subplots(figsize=(8, 4))
ax.plot(df['date'], df['pm25'])
ax.set_title('PM2.5 over time')
ax.set_xlabel('Date'); ax.set_ylabel('PM2.5 (µg/m³)')
```

## L8.2 — Seaborn `[McKinney Ch9 p.310–325]`

```python
import seaborn as sns
h = pd.read_csv('housing_prices.csv')
sns.scatterplot(data=h, x='median_income', y='median_house_value',
                hue='ocean_proximity', alpha=0.4, s=10)
```

## L8.5 — CVD-safe palettes `[Expert]`

```python
# Use a CVD-friendly palette
sns.set_palette('colorblind')
# Never red-green for primary contrast
```

## L8.7 — Plotly Express `[Expert]`

```python
import plotly.express as px
fig = px.scatter(h, x='median_income', y='median_house_value',
                 color='ocean_proximity', opacity=0.4)
fig.show()
```

## L8.8 — Matplotlib advanced `[McKinney Ch9 p.290–310]`

```python
fig, ax = plt.subplots(figsize=(8, 5))
ax.plot(df['date'], df['pm25'], linewidth=1.5)
ax.set_xticks(df['date'][::max(len(df)//8, 1)])
ax.tick_params(axis='x', rotation=45)
ax.legend(['PM2.5'], loc='upper left', frameon=False)
fig.tight_layout()
# fig.savefig('outputs/m8_pm25.pdf', dpi=300, bbox_inches='tight')
```
